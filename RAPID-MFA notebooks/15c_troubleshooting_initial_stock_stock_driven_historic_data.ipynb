# RAPID-MFA 15c: Trouble shooting: initial stock with a stock driven model

## Descriptions

This variant of the stock driven model allows to solve the _initial stocks problem_, which represents any stocks that are already existing in the system prior to the time series, because the model assumes that the initial inflow starts building up the stock from zero. 

This solution uses _simple linear interpolation_ to estimate stock prices before 1950, as no historical stock data is available. This example assumes the stock was zero in 1871.

### Exogenous variables
__Stock:__ Railway data of Japan from _Tanikawa, H., Fishman, T., Okuoka, K., & Sugimoto, K. (2015). The weight of society over time and space: A comprehensive account of the construction material stock of Japan, 1945–2010. Journal of Industrial Ecology, 19(5), 778-791._

__Survival curve:__ mock values. Normal distribution. chosen to exemplify that the model works even in a case in which survival in year 0 is not 100%, i.e. there is already some outflow (throughput) from each inflow in its year of occurance.

__Extended input data:__ Uses the inital stock and assumes a liner build up of the stock from 1871 to 1950 to not treat the inital stock as one cohort. The choice of linear interpolation is arbitary and could be replaced with other forms of interpolations.

## 1. Load libraries

```python
# %% 1. Load libraries
import numpy as np
import pandas as pd
import scipy.stats
```

## 2. Define example input data

```python
# %% 2. Define example input data
input_data = pd.Series([3648714.685, 3650304.605, 3652889.659, 3655055.855, 3655772.411, 3662912.35, 
                        3663887.175, 3670963.548, 3678930.998, 3668618.801, 3672909.848, 3674826.182,
                        3664671.192, 3673137.674, 3799949.102, 3792669.285, 3790034.365, 3777331.215,
                        3769431.593, 3753899.973, 3751656.137, 3733914.943, 3750868.872, 3768687.714,
                        3779471.855, 3867815.975, 3869118.191, 3875078.363, 3879239.156, 3883067.295,
                        3885778.279, 3898386.394, 4079394.532, 4077609.408, 4069241.351, 4057790.434,
                        4052450.883, 3979092.586, 4007890.682, 3967878.751, 3972743.035, 3979799.233,
                        3977730.399, 3979281.08, 3987864.764, 3976941.809, 3985063.413, 4028347.781,
                        4033346.479, 4037376.746, 4045015.573, 4043855.96, 4064250.081, 4061856.096,
                        4082347.234, 4089728.633, 4080102.939, 4071482.315, 4074341.147, 4075275.163,
                        4099241.12, 4131764.257], 
                        name = "inflows",
                        index = [1950, 1951, 1952, 1953, 1954, 1955, 1956, 1957, 1958, 1959,
                                1960, 1961, 1962, 1963, 1964, 1965, 1966, 1967, 1968, 1969,
                                1970, 1971, 1972, 1973, 1974, 1975, 1976, 1977, 1978, 1979,
                                1980, 1981, 1982, 1983, 1984, 1985, 1986, 1987, 1988, 1989,
                                1990, 1991, 1992, 1993, 1994, 1995, 1996, 1997, 1998, 1999,
                                2000, 2001, 2002, 2003, 2004, 2005, 2006, 2007, 2008, 2009,
                                2010, 2011]) 

# Parameters for the survival curve
ages = range(0, len(input_data))
mean = 30
std_dev = 10

# create the survival curve
survival_curve = scipy.stats.norm.sf(ages, loc=mean, scale=std_dev)

#Define the extended input data for cell 5
# Extending the stocks backwards in time
extended_input_data = pd.DataFrame(index=range(1871, 2012, 1))
extended_input_data.loc[1950:, 'stocks'] = input_data
extended_input_data.loc[1871, 'stocks'] = 0

#Interpolate the missing values in the 'stocks' column
extended_input_data['stocks'] = extended_input_data['stocks'].interpolate()

ages_extended = range(0, len(extended_input_data))
survival_curve_extended = scipy.stats.norm.sf(ages_extended, loc=mean, scale=std_dev)
```

## 3. Run the stock-driven model

The steps from hereon are a repetition of notebook 2, but with different input data. If you want a more detailed account on how to do this, see notebook 2.

```python
# %% 3. Run the stock-driven model
units = {'time_interval':   'year', 
         'flows':           'kg/year', 
         'stocks':          'kg'}

timeframes = cohorts = input_data.index  # timeframes are the "snapshots" of the system in every time interval, while cohorts are the inflows that enter the system in every time interval. They are the same because there is one inflow per time interval.
time_interval = 1
cohort_stop = (cohorts[-1] + time_interval)  # because of exclusive slicing in python, an extra time interval is required
mfa_system = pd.DataFrame(index=timeframes, columns = ['stocks', 
                                                         'inflows', 
                                                         'outflows', 
                                                         'stock_changes'])

mfa_system['stocks'] = input_data

# create survival curves matrix filled with zeros
survival_curves_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# and populate that matrix
for cohort in cohorts:
    # at each iteration, shift the survival curve and fill the next column of the survival curves matrix with the shifted survival curve from the diagonal downwards.
    survival_curves_matrix.loc[cohort: , cohort] = survival_curve[0 : cohort_stop - cohort]
   
# initialize the cohort_survival_matrix filled with zeros
cohorts_survival_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# And populate it
for cohort in cohorts:
    # calculate the inflow of the time-th timeframe
    mfa_system.loc[cohort, 'inflows'] = (
        mfa_system.loc[cohort, 'stocks'] - cohorts_survival_matrix.loc[cohort, : cohort - 1].sum()
    ) / survival_curves_matrix.loc[cohort, cohort]
    # now that the time-th inflow is known, scale the shifted survival curve with this inflow
    cohorts_survival_matrix.loc[:, cohort] = survival_curves_matrix.loc[:, cohort] * mfa_system.loc[cohort, 'inflows']
    
mfa_system["stock_changes"] = np.diff(mfa_system["stocks"], prepend=0)
mfa_system["outflows"] = mfa_system["inflows"] - mfa_system["stock_changes"]
```

## 4. Identify the initial stock issue
To identify the initial stocks before the year 1950, we can visualize the stocks by cohorts

```python
# plot stocks by cohorts
ax = cohorts_survival_matrix.plot(kind="area", stacked=True, legend=False, lw=0, ylabel="stocks [kg]")
```

## 5. Recalculate the MFA system with the extended interpolated values

```python
# %% 5. Recalculate the MFA system with the extended input data
timeframes = cohorts = extended_input_data.index  # timeframes are the "snapshots" of the system in every time interval, while cohorts are the inflows that enter the system in every time interval. They are the same because there is one inflow per time interval.
cohort_stop = (cohorts[-1] + time_interval)  # because of exclusive slicing in python, an extra time interval is required


# create an empty dataframe to hold the values of the MFA system
mfa_system = pd.DataFrame(index=timeframes, columns = ['stocks', 
                                                        'inflows', 
                                                        'outflows', 
                                                        'stock_changes'])

mfa_system['stocks'] = extended_input_data

# create survival curves matrix filled with zeros
survival_curves_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# and populate that matrix
for cohort in cohorts:
    # at each iteration, shift the survival curve and fill the next column of the survival curves matrix with the shifted survival curve from the diagonal downwards.
    survival_curves_matrix.loc[cohort: , cohort] = survival_curve_extended[0 : cohort_stop - cohort]
   
# initialize the cohort_survival_matrix filled with zeros
cohorts_survival_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# And populate it
for cohort in cohorts:
    # calculate the inflow of the time-th timeframe
    mfa_system.loc[cohort, 'inflows'] = (
        mfa_system.loc[cohort, 'stocks'] - cohorts_survival_matrix.loc[cohort, : cohort - 1].sum()
    ) / survival_curves_matrix.loc[cohort, cohort]
    # now that the time-th inflow is known, scale the shifted survival curve with this inflow
    cohorts_survival_matrix.loc[:, cohort] = survival_curves_matrix.loc[:, cohort] * mfa_system.loc[cohort, 'inflows']

mfa_system["stock_changes"] = np.diff(mfa_system["stocks"], prepend=0)
mfa_system["outflows"] = mfa_system["inflows"] - mfa_system["stock_changes"]

mfa_system_with_units = mfa_system.rename(columns={"stocks" : 'stocks [' + units['stocks'] + ']', 
                                                    "inflows" : 'inflows [' + units['flows'] + ']', 
                                                    "outflows" : 'outflows [' + units['flows'] + ']', 
                                                    "stock_changes" : 'stock_changes [' + units['flows'] + ']'})
```

```python
ax = cohorts_survival_matrix.loc[1950:,:].plot(kind="area", stacked=True, legend=False, lw=0, ylabel="stocks [kg]")
```

## 6. Concluding remarks
This simple linear interpolation introduced the _sudden jolt problem_ in the pre-1950 data. While this could be addressed  using the methods in `RAPID-MFA-16: Sudden jumps`, it might be irrelevant since the focus of this example focus was on data from 1950 onward. This notebook reconstructed a stock composition for 1950, and the inflows and outflows appear reasonable.
