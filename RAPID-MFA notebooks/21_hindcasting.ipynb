This notebook generalizes the hindcasting model of [Böcher et al., 2025](https://doi.org/10.1016/j.resconrec.2025.108198). 

The hindcasting model takes two inputs:
- A _current stocks_ in a single timeframe with its cohorts' age composition (e.g. current stocks' cohorts by construction year)
- A _survival curve_ describing the survival/depreciation of inflow cohorts in the stocks as time goes/went by

And produces one main output:
- The _inflows_ time series

Which in turn feed into a _flow driven model_ that produces two main model outputs:

- The _stocks_ time series
- The _outflows_ time series

And two auxiliary model outputs:
- The _cohorts survival matrix_ table
- The _stock changes_ (net addition to stock) time series

### Hindcasting the historical inflows

The hindcasting model starts from a stock in a single timeframe $t$ that has a known cohorts composition:

$
\textcolor{#DC322F}{\text{stocks}(t)} = \sum_{c=t_0}^{t} \text{cohorts}(c)
$

This is equivalent to a given _row_ $t$ in a dynamic MFA's _cohorts survival matrix_. Reminder: the core flow-driven model creates rows in the _cohorts survival matrix_ uses the equation:

$
\textcolor{#DC322F}{\text{stocks}(t)} = \sum_{c=t_0}^{t}{ [ \textcolor{#2A8596}{\text{inflows}(c)} \times \textcolor{#722299}{\text{survival\_curve}(t-c)}  ] } 
$

Combining these two equations, each of the stock’s cohorts is equal to the still-surviving inflow from a past timeframe:

$\text{cohorts}(c) = \textcolor{#2A8596}{\text{inflows}(c)} \times \textcolor{#722299}{\text{survival\_curve}(t-c)}$

Which enables the calculation of the original value of each inflow that created each cohort:

$\textcolor{#2A8596}{\text{inflows}(c)} = \frac{\text{cohorts}(c)}{\textcolor{#722299}{\text{survival\_curve}(t-c)}}$

This results in a full $\text{inflows}$ time series from the oldest cohort $c$ to the existing stock's timeframe $t$. This inflow can then be used in a _flow driven dynamic MFA_ model to fill in the entire MFA system of historical stocks and flows.

## 3. Set up the MFA system

This code cell is identical to the core MFA models, except that the time variables _timeframes_ and _cohorts_ are defined by the number (and timeframes) of the cohorts composing the _known stock_.

In other words, the time period that this MFA system will cover is from the inflow of the earliest cohort in the stock (in this example 1950) until the inflow of the timeframe of the known stock (in this example 2019).

```python
# %% 3. Set up the MFA system
units = {'time_interval': 'year', 
         'flows':         'number of buildings/year', 
         'stocks':        'number of buildings'}

timeframes = cohorts = known_stocks.index  # timeframes are the "snapshots" of the system in every time interval, while cohorts are the inflows that enter the system in every time interval. They are the same because we have one inflow per time interval.

time_interval = 1 # Interval between timeframes, often 1 year or 1 month etc.

cohort_stop = (cohorts[-1] + time_interval)  # because of exclusive slicing in python, an extra time interval is required

mfa_system = pd.DataFrame(index = timeframes, columns = ['stocks',
                                                         'inflows', 
                                                         'outflows', 
                                                         'stock_changes'])
```

## 4. Create the survival curves matrix

In this example, pre-1990 cohorts have a different survival curve than post-1990 cohorts, so this code cell is more similar to `RAPID-MFA-6: Varying lifetimes`. If all cohorts have the same survival curve, then the regular code from the core models can be used.

```python
# %% 4. Create survival curves matrix
# create survival curve matrix filled with zeros
survival_curves_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# and populate that matrix
for cohort in cohorts:
    # at each iteration, we shift the survival curve and we fill the survival curve matrix with the shifted survival curve
    if cohort <= 1990:     
        survival_curves_matrix.loc[cohort: , cohort] = survival_curve_pre_1990[0 : cohort_stop - cohort]
    else:
        survival_curves_matrix.loc[cohort: , cohort] = survival_curve_post_1990[0 : cohort_stop - cohort]

# and visualize the survival_curves_matrix matrix with a heatmap
sns.heatmap(survival_curves_matrix, annot=False).set(xlabel ="Cohorts", ylabel = "Timeframe")
```

## 5. Hindcasting past inflows

$\text{inflows}(c) = \frac{\text{cohorts}(c)}{\text{survival\_curve}(t - c)}$


```python
for cohort in cohorts:
    mfa_system.loc[cohort, "inflows"] = known_stocks[cohort] / survival_curves_matrix.loc[known_stocks_timeframe, cohort]
```

## 6. Run the flow-driven model

Now that the historical inflows are known, this code cell is identical to `RAPID-MFA-1: The flow driven model`

```python
# %% Create cohorts survival matrix

# initialize the cohort_survival_matrix filled with zeros
cohorts_survival_matrix = pd.DataFrame(0, index=timeframes, columns=cohorts, dtype=float)

# And populate it
for cohort in cohorts:
    # scale the shifted survival curve with the inflow of the time-th timeframe
    cohorts_survival_matrix.loc[:, cohort] = survival_curves_matrix.loc[:, cohort] * mfa_system['inflows'].loc[cohort]

# %% Calculate the MFA system
# %%% Calculate stocks 
# the stocks are the sum of the surviving cohorts, which is the sum of cohorts survival matrix row by row.
mfa_system["stocks"] = cohorts_survival_matrix.sum(axis='columns')

# %%% calculate stock changes (net change to stock)
# stock_changes is the difference between two timeframes.
# prepend=0 replicates the convention that stock_change is the difference between current timeframe and previous timeframe. 
mfa_system["stock_changes"] = np.diff(mfa_system["stocks"], prepend=0)

# %%% Calculate outflows
# using mass balance
mfa_system["outflows"] = mfa_system["inflows"] - mfa_system["stock_changes"]
```

## 7. Diagnostics
The flow-driven model is "blind" to the fact that its inflows were hindcasted.

Simple comparisons can therefore confirm that the flow-driven model manages to accurately recreate the "known stock" and its composition:

```python
# the total known stock is the sum of its cohorts
known_stocks_total = known_stocks.sum()

# compare the total stock of the last timeframe to the originally known stock
mfa_system.loc[known_stocks_timeframe, "stocks"] == known_stocks_total
```

```python
# compare each cohort of the last timeframe to the originally known stock's cohorts
cohorts_survival_matrix.loc[known_stocks_timeframe, :] == known_stocks
```
